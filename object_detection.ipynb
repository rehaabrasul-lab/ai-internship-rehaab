```python
import supervision as sv
from rfdetr import RFDETRMedium
from rfdetr.assets.coco_classes import COCO_CLASSES

model = RFDETRMedium()
detections = model.predict("https://media.roboflow.com/dog.jpg", threshold=0.5)
labels = [f"{COCO_CLASSES[class_id]}" for class_id in detections.class_id]

model.inference(compile=False, inplace=True, dtype="float16")

annotated_image = sv.BoxAnnotator().annotate(detections.metadata["source_image"], detections)
annotated_image = sv.LabelAnnotator().annotate(annotated_image, detections, labels)
sv.plot_image(annotated_image[:, :, ::-1])
```

```python
import supervision as sv
from rfdetr import RFDETRMedium
from rfdetr.assets.coco_classes import COCO_CLASSES
from PIL import Image, ImageOps
import requests
from io import BytesIO

model = RFDETRMedium()

url = "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTvbrs2clZGG5YkdR9uqtITXEKT036ycd-kBVyr5QXmo14z65VM8bK53cgc&s=10"
image = Image.open(BytesIO(requests.get(url).content))

# Add 100px padding
image = ImageOps.expand(image, border=20, fill="white")

detections = model.predict(image, threshold=0.5)

labels = [COCO_CLASSES[i] for i in detections.class_id]

image = sv.BoxAnnotator().annotate(image, detections)
image = sv.LabelAnnotator(
    text_position=sv.Position.TOP_LEFT
).annotate(image, detections, labels)

sv.plot_image(image)
```
